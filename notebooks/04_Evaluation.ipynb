```python
from numpy import load
from json import load as json_load
from src import evaluate_model, plot_confusion_matrix, plot_prediction_examples, predict_image
from src.config import PROCESSED_DATA_DIR, REPORTS_FIGURES_DIR, FINAL_MODEL_PATH, RAW_DATA_DIR
```

```python
from tensorflow.keras.models import load_model

model = load_model(FINAL_MODEL_PATH)
x_test = load(PROCESSED_DATA_DIR / "x_test.npy")
y_test = load(PROCESSED_DATA_DIR / "y_test.npy")
with open(PROCESSED_DATA_DIR / "class_names.json") as f:
    class_names = json_load(f)["class_names"]
print(f"Test: {x_test.shape}, classes: {class_names}")
```

```python
eval_result = evaluate_model(model, x_test, y_test, class_names)
```

```python
plot_confusion_matrix(eval_result["y_true"], eval_result["y_pred"], class_names, REPORTS_FIGURES_DIR)
plot_prediction_examples((x_test * 255).astype("uint8"), eval_result["y_true"], eval_result["y_pred"], class_names, REPORTS_FIGURES_DIR)
```
